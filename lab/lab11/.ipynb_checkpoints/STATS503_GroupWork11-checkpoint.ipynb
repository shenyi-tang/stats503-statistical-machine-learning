### **Problem 2: Gradient Descent with Autograd** (2 points)

Consider the simple regression problem given $X=(x_1,\dots,x_n)$ as a vector of length $n$ and targets $Y=(y_1,\dots,y_n)$. Use gradient descent to minimize the Mean Square Error loss

$$f(w) = \frac{1}{n}\sum_i (y_i - w_0 - w_1 x_i)^2.$$

Write a function which takes as input $X$, $Y$, $\eta$ (step size), $\epsilon$ (convergence threshold) and `max_iter` (terminate if number of iterations exceed this). Consider the algorithm converged when
$$\Vert w^{(t+1)} - w^{(t)} \Vert < \epsilon.$$

**NOTE**: This is the same problem we did last week in groupwork but we used `numpy` to accomplish it. Please use `torch` this week.

```python
def mse_gd(X, y, eta=1e-1, epsilon=1e-6, max_iter=1000):
    """
    X: torch tensor of shape (n,)
    y: torch tensor of shape (n,)
    eta: float, learning rate
    epsilon: float, convergence threshold
    max_iter: int, maximum number of iterations

    returns: torch tensor of shape (2,) containing the learned parameters w = [w0, w1]
    """

    # Ensure float type
    X = X.float()
    y = y.float()
    n = X.shape[0]

    # Add bias term (column of ones)
    X_design = torch.stack([torch.ones(n), X], dim=1)  # shape (n, 2)

    # Initialize parameters with gradients
    w = torch.randn(2, requires_grad=True)

    for i in range(max_iter):
        # Forward pass: compute predictions
        y_pred = X_design @ w

        # Compute Mean Squared Error loss
        loss = torch.mean((y - y_pred) ** 2)

        # Backward pass: compute gradients
        loss.backward()

        # Gradient descent step
        with torch.no_grad():
            w_new = w - eta * w.grad

            # Check convergence
            if torch.norm(w_new - w) < epsilon:
                w = w_new.detach().clone().requires_grad_(True)
                break

            # Update weights
            w.copy_(w_new)
            w.requires_grad_(True)

        # Clear gradients
        w.grad.zero_()

    return w.detach()
```

### **Problem 4: Training a Model** (4 points)

We have provided a commented `train_model` function for you to infill with descriptions of each argument.

Procede then to create a small neural network with one hidden layer of width 32 and, using the training function you infilled, train it for 100 epochs.


```python
import random
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)

def train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs=100, patience=0) -> tuple[list, list, nn.Module]:
    """
      model (torch.nn.Module): The neural network
      train_loader (torch.utils.data.DataLoader): Training data
      val_loader (torch.utils.data.DataLoader): Validation data
      criterion (torch.nn Loss function): The loss function
      optimizer (torch.optim): The optimizer used to make gradient steps
      num_epochs (int): The number of training epochs to perform
      patience (int): Number of epochs to use for early stopping. If 0, early stopping is not used.
    """

    training_losses = []
    validation_losses = []
    min_validation_loss = float('inf')
    patience_counter = 0

    for epoch in range(num_epochs):
        model.train()  # Set model to training mode
        train_loss = 0

        # Training loop
        for batch_X, batch_y in train_loader:
            optimizer.zero_grad()              # Clear gradients
            outputs = model(batch_X)           # Forward pass
            loss = criterion(outputs, batch_y) # Compute loss
            loss.backward()                    # Backward pass
            optimizer.step()                   # Update parameters
            train_loss += loss.item()

        train_loss /= len(train_loader)
        training_losses.append(train_loss)

        # Validation loop
        model.eval()  # Set model to evaluation mode
        val_loss = 0
        with torch.no_grad():
            for batch_X, batch_y in val_loader:
                outputs = model(batch_X)
                loss = criterion(outputs, batch_y)
                val_loss += loss.item()
        val_loss /= len(val_loader)
        if patience:
          if val_loss < min_validation_loss:
              min_validation_loss = val_loss
              patience_counter = 0
          else:
              patience_counter += 1
              if patience_counter >= patience:
                  print("Early stopping triggered")
                  break
        validation_losses.append(val_loss)

        # Print progress every 10 epochs
        if epoch % 10 == 0:
            print(f"Epoch {epoch:3d} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")


    return training_losses, validation_losses, model
```

```python
# Write your model here. # HINT: Maybe pytorch has an object that connects layers together in a block?
model = nn.Sequential(
    nn.Linear(X_train.shape[1], 32),  # Hidden layer with 32 units
    nn.ReLU(),                 # Activation function
    nn.Linear(32, 2) # Output layer with 2 units
)

# Loss function here. # HINT: Is this a regression or classification problem?

# For a classification problem, we use CrossEntropyLoss.
criterion = nn.CrossEntropyLoss()

# Use the Adam optimizer.
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Train the model for 100 epochs.
training_losses, validation_losses, model = train_model(model, train_loader, validation_loader, criterion, optimizer, num_epochs=100)
```

```python
# CHECK HOW THE MODEL DID
# NO CHANGES NECESSARY HERE
plt.plot(training_losses, label='Training Loss')
plt.plot(validation_losses, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
# def build_model(n_hidden=1, n_neurons=30, activation='relu', input_dim=0, output_dim=0, include_dropout=False):
#     activation_map = {
#         'relu': nn.ReLU,
#         'tanh': nn.Tanh,
#         'sigmoid': nn.Sigmoid,
#         'leaky_relu': nn.LeakyReLU,
#     }

#     return None

import torch.nn as nn

def build_model(n_hidden=1, n_neurons=30, activation='relu', input_dim=0, output_dim=0, include_dropout=False):
    activation_map = {
        'relu': nn.ReLU,
        'tanh': nn.Tanh,
        'sigmoid': nn.Sigmoid,
        'leaky_relu': nn.LeakyReLU,
    }

    if activation not in activation_map:
        raise ValueError(f"Unsupported activation: {activation}. Choose from {list(activation_map.keys())}.")

    act_fn = activation_map[activation]
    layers = []

    # Input layer
    layers.append(nn.Linear(input_dim, n_neurons))
    layers.append(act_fn())

    # Hidden layers
    for _ in range(n_hidden - 1):
        layers.append(nn.Linear(n_neurons, n_neurons))
        layers.append(act_fn())
        if include_dropout:
            layers.append(nn.Dropout(p=0.5))  # Dropout added conditionally

    # Output layer
    layers.append(nn.Linear(n_neurons, output_dim))

    return nn.Sequential(*layers)
```

## **Problem 6: Adding regularization**

### **Problem 6a: Weight Decay** (2 points)




---


You may be familiar with Occam's Razor principle: given two explanations for something, the explanation most likely to be correct is the "simplest" one, the one that makes the least amount of assumptions. This also applies to the models learned by neural networks: given 2 networks that achieve a similar performance, go with the simpler one! Less risk of overfitting!

* [L1 regularization](https://developers.google.com/machine-learning/glossary/#L1_regularization), where the cost added is proportional to the absolute value of the weights coefficients (i.e. to what is called the "L1 norm" of the weights). L1 regularization produces sparse models because a decrease of 0.1 -> 0 is as penalty reducing as 10.1 -> 10.0 .

* [L2 regularization](https://developers.google.com/machine-learning/glossary/#L2_regularization), where the cost added is proportional to the square of the value of the weights coefficients (i.e. to what is called the squared "L2 norm" of the weights). L2 regularization is also called **weight decay** in the context of neural networks.

Here, we apply L2 regularization using the `weight_decay` feature of PyTorch.
Feel free to read the [article](https://pytorch.org/docs/stable/generated/torch.optim.Adam.html) and adjust the `weight_decay` parameter to remove overfitting in the medium size model.

### **Problem 6b: Dropout** (2 points)


---


The intuitive explanation for dropout is that because individual nodes in the network should not be able to rely on the full output of the previous layer. Not all signal is useful signal.

Dropout, applied to a layer, consists of randomly "dropping out" (i.e. set to zero) a number of output features of the layer during training. For example, a given layer would normally have returned a vector `[0.2, 0.5, 1.3, 0.8, 1.1]` for a given input sample during training; after applying dropout, this vector will have a few zero entries distributed at random, e.g. `[0, 0.5, 1.3, 0, 1.1]`.

The "dropout rate" is the fraction of the features that are being zeroed-out; it is usually set between 0.2 and 0.5. At test time, no units are dropped out, and instead the layer's output values are scaled down by a factor equal to the dropout rate, so as to balance for the fact that more units are active than at training time.

You usually want to keep drop out low near the input (the input is what we believe is important to learn from) and higher as you go deeper (not all learned functions of the input are useful).



---

Update your `build_model` function to include dropout layers in accordance with the number of layers. We can pass in a list of the correct size to `include_dropout` and add Dropout norms as needed. Build a model and train it. Then plot validation and training losses.

```python
# Update the build_model function to include dropout
def build_model(n_hidden=5, n_neurons=16, activation='relu', input_dim=0, output_dim=0, include_dropout=False):
    activation_map = {
        'relu': nn.ReLU,
        'tanh': nn.Tanh,
        'sigmoid': nn.Sigmoid,
        'leaky_relu': nn.LeakyReLU,
    }

    if activation not in activation_map:
        raise ValueError(f"Unsupported activation: {activation}. Choose from {list(activation_map.keys())}.")

    act_fn = activation_map[activation]
    layers = []

    # Input layer
    layers.append(nn.Linear(input_dim, n_neurons))
    layers.append(act_fn())
    if include_dropout:
        layers.append(nn.Dropout(p=0.5))

    # Hidden layers
    for _ in range(n_hidden - 1):
        layers.append(nn.Linear(n_neurons, n_neurons))
        layers.append(act_fn())
        if include_dropout:
            layers.append(nn.Dropout(p=0.5))

    # Output layer
    layers.append(nn.Linear(n_neurons, output_dim))

    return nn.Sequential(*layers)

# Build a dropout model and train it
dropout_model = build_model(n_hidden=5, n_neurons=32, activation='relu',
                            input_dim=X_train.shape[1], output_dim=2,
                            include_dropout=True).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(dropout_model.parameters(), lr=0.001)

training_losses_dropout, val_losses_dropout, trained_model_dropout = train_model(
    dropout_model, train_loader, validation_loader, criterion, optimizer, num_epochs=100
)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(training_losses_dropout, label='Training Loss')
plt.plot(val_losses_dropout, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training vs Validation Loss with Dropout')
plt.legend()
plt.grid(True)
plt.show()
```

### **Problem 6c: Early Stopping** (3 points)



---

Of course, sometimes the change in loss from epoch to epoch can by noisy, so we employ something called  "patience". The patience simply means how many epochs *without* a decrease in validation loss we need to observe to stop training.



---


Update your `train_model` function to have a patience argument for early stopping. You may implement early stopping using Pytorch classes or with a manual implementation, whichever you find easiest.
