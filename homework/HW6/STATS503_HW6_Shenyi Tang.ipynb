**It was mentioned in this chapter that a cubic regression spline with one knot at $\xi$ can be obtained using a basis of the form $x,\ x^2,\ x^3,\ (x - \xi)^3_+$, where $(x - \xi)^3_+ = (x - \xi)^3$ if $(x > \xi)$ and equals 0 otherwise. We will now show that a function of the form 
$$f(x) = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \beta_4 (x - \xi)^3_+$$
is indeed a cubic regression spline, regardless of the values of $\beta_0, \beta_1, \beta_2, \beta_3, \beta_4$.**

**(a) Find a cubic polynomial
    $$
    f_1(x) = a_1 + b_1 x + c_1 x^2 + d_1 x^3
    $$
    such that $f(x) = f_1(x)$ for all $x \leq \xi$. Express $a_1, b_1, c_1, d_1$ in terms of $\beta_0, \beta_1, \beta_2, \beta_3, \beta_4$.**

$$
\begin{align*}
a_1=\beta_0 \\
b_1=\beta_1 \\
c_1=\beta_2 \\
d_1=\beta_3
\end{align*}
$$

***

**(b) Find a cubic polynomial
    $$
    f_2(x) = a_2 + b_2 x + c_2 x^2 + d_2 x^3
    $$
    such that $f(x) = f_2(x)$ for all $x > \xi$. Express $a_2, b_2, c_2, d_2$ in terms of $\beta_0, \beta_1, \beta_2, \beta_3, \beta_4$.**

$$
\begin{align*}
    a_2 &= \beta_0-\beta_4\xi^3 \\
    b_2 &= \beta_1+3\beta_4\xi^2 \\
    c_2 &= \beta_2-3\beta_4\xi \\
    d_2 &= \beta_3+\beta_4
\end{align*}
$$

***

**(c) Show that $f_1(\xi) = f_2(\xi)$. That is, $f(x)$ is continuous at $\xi$.**

$$
\begin{align*}
f_1(\xi) &= \beta_0 + \beta_1\xi + \beta_2\xi^2 + \beta_3\xi^3 \\
f_2(\xi) &= \beta_0-\beta_4\xi^3 + (\beta_1+3\beta_4\xi^2)\xi + (\beta_2-3\beta_4\xi)\xi^2 + (\beta_3+\beta_4)\xi^3 \\
&= \beta_0 + \beta_1\xi + \beta_2\xi^2 + \beta_3\xi^3
\end{align*}
$$
That is $f(x)$ is continous at $\xi$

***

**(d) Show that $f'_1(\xi) = f'_2(\xi)$. That is, $f'(x)$ is continuous at $\xi$.**

$$
\begin{align*}
f'_1(\xi) &= \beta_1 + 2\beta_2\xi + 3\beta_3\xi^2 \\
f'_2(\xi) &= (\beta_1+3\beta_4\xi^2) + 2(\beta_2-3\beta_4\xi)\xi + 3(\beta_3+\beta_4)\xi^2 \\
&= \beta_1 + 2\beta_2\xi + 3\beta_3\xi^2
\end{align*}
$$
That is, $f'(x)$ is continuous at $\xi$

***

**(e) Show that $f''_1(\xi) = f''_2(\xi)$. That is, $f''(x)$ is continuous at $\xi$**

$$
\begin{align*}
f''_1(\xi) &= 2\beta_2 + 6\beta_3\xi \\
f''_2(\xi) &= 2(\beta_2-3\beta_4\xi) + 6(\beta_3+\beta_4)\xi \\
&= 2\beta_2 + 6\beta_3\xi
\end{align*}
$$
That is, $f''(x)$ is continuous at $\xi$

#### Question 7

**This question uses the following variables from the `boston` data, which can be downloaded from canvas. We are interested in predicting the air quality variable `nox`.**

**(a) For this problem, we will estimate the conditional expectation of `nox` given `indus` (one predictor). We will consider three different estimators:**
- **Linear regression with spline features and knots at {0,12,17,30}**
- **Linear regression with polynomial features of degree 12**
- **K-nearest-neighbors regression with 13 neighbors**

*Estimate the mean squared predictive error for each estimator using 4-fold cross-validation (remember to use the same splits for each estimator). Report the error for each of the estimators. Which estimator gives the lowest error?*

*Fit each estimator to the entire dataset and plot the three predicted curves against a scatterplot of `indus` vs `nox` in the data. Set `alpha=0.1` in the scatterplot. (Otherwise, you will not be able to tell that there are many points that are “on top of each other”). Why might one choose to highlight the curve with the highest error when discussing air quality in a public presentation? A short explanation of one to three sentences will suffice.*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, SplineTransformer
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
```

```python
# read data
bos = pd.read_csv("boston.csv", index_col=0)
bos.head()
```

```python
X = bos['indus'].values.reshape(-1,1)
y = bos['nox']

knots = np.array([[0], [12], [17], [30]])
kf = KFold(n_splits=4, shuffle=True, random_state=42)

# transformer
spline = SplineTransformer(knots=knots)
poly = PolynomialFeatures(degree=12)

mse_spline = []
mse_poly = []
mse_knn = []

for i, (train_idx, test_idx) in enumerate(kf.split(X)):
    X_train = X[train_idx]
    X_test = X[test_idx]
    y_train = y[train_idx]
    y_test = y[test_idx]

    # linear regression with spline features
    X_train_spline = spline.fit_transform(X_train)
    X_test_spline = spline.fit_transform(X_test)
    m1 = LinearRegression()
    m1.fit(X_train_spline, y_train)
    y_pred_spline = m1.predict(X_test_spline)
    mse_spline.append(mean_squared_error(y_test, y_pred_spline))

    # linear regression with polynomial features 
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.fit_transform(X_test)
    m2 = LinearRegression()
    m2.fit(X_train_poly, y_train)
    y_pred_poly = m2.predict(X_test_poly)
    mse_poly.append(mean_squared_error(y_test, y_pred_poly))

    # knn regression
    m3 = KNeighborsRegressor(n_neighbors=13)
    m3.fit(X_train, y_train)
    y_pred_knn = m3.predict(X_test)
    mse_knn.append(mean_squared_error(y_test, y_pred_knn))

result = {"mse_spline_avg": np.mean(mse_spline),
         "mse_poly_avg": np.mean(mse_poly),
         "mse_knn_avg": np.mean(mse_knn)}

print(result)

est = min(result.items(), key=lambda x: x[1])
print(f"{est[0]} has the lowest mse")
```

```python
X_spline = spline.fit_transform(X)
X_poly = poly.fit_transform(X)

m1.fit(X_spline, y)
m2.fit(X_poly, y)
m3.fit(X, y)

X_plot = np.linspace(X.min(), X.max(), 300).reshape(-1, 1)

X_plot_spline = spline.transform(X_plot)
X_plot_poly = poly.transform(X_plot)

y_plot_spline = m1.predict(X_plot_spline)
y_plot_poly = m2.predict(X_plot_poly)
y_plot_knn = m3.predict(X_plot)

plt.figure(figsize=(10, 6))
sns.scatterplot(x=X.flatten(), y=y, alpha=0.1, label="Observed", color="black")

plt.plot(X_plot, y_plot_spline, label="Spline Regression", color="blue")
plt.plot(X_plot, y_plot_poly, label="Polynomial Regression (deg=12)", color="red", linestyle="dashed")
plt.plot(X_plot, y_plot_knn, label="KNN Regression (k=13)", color="green", linestyle="dotted")

plt.xlabel("indus")
plt.ylabel("nox")
plt.legend()
plt.title("Regression Models for Predicting NOx Levels")
plt.show()
```

Highlighting the curve with the highest error can illustrate issues like overfitting or underfitting. It demonstrates the importance of proper model selection, feature representation, and robustness in predictive modeling. This helps convey why certain models fail to generalize well and the necessity of balancing bias and variance.

**(b) Use `pygam` to predict `nox` from `dis`, `indus`, and `rad`. Treat each of these variables as continuous.**
- **Use `pygam` to fit the data, considering different choices for the `lam` that controls the regularization strength. For each feature, consider four choices of regularization, $\lambda \in \{0.5,1,5,10\}$. (There will be $4^3 = 64$ fits in total.) Use `pygam`’s gridsearch method to identify the best choice. Report the best $\lambda$ value found.**
- **Plot the partial dependence for each of the three features.**
- **The feature `rad` indicates an index of accessibility to highways. How well do you think the estimated partial dependence for this feature represents the true relationship between `rad` and `nox`? Explain.**

```python
from pygam import LinearGAM, s, f
from sklearn.model_selection import GridSearchCV

# Define the feature matrix and response variable
X = bos[['dis', 'indus', 'rad']].values
y = bos['nox'].values

# Define the grid of lambda values
lam_values = [0.5, 1, 5, 10]
lam_grid = np.array(np.meshgrid(lam_values, lam_values, lam_values)).T.reshape(-1, 3)  # 4^3 = 64 combinations

# Fit the GAM model using grid search
gam = LinearGAM(s(0) + s(1) + s(2))
grid = gam.gridsearch(X, y, lam=lam_grid)

# Get the best lambda
best_lam = gam.lam
print(f"Best lambda values found: {best_lam}")

plt.figure(figsize=(15, 5))

for i, feature in enumerate(['dis', 'indus', 'rad']):
    plt.subplot(1, 3, i + 1)
    XX = grid.generate_X_grid(term=i)
    plt.plot(XX[:, i], grid.partial_dependence(term=i, X=XX))
    plt.plot(XX[:, i], grid.partial_dependence(term=i, X=XX, width=0.95)[1], c='r', ls='--')
    plt.title(f'{feature}')

plt.tight_layout()
plt.show()
```

The rad subplot shows an unclear relationship with nox, suggesting that rad may not have a strong effect. The wide confidence intervals indicate high uncertainty, likely due to limited data in some areas.
