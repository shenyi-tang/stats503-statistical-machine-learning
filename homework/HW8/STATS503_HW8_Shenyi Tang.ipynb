#### Question 2

**ISLP chapter 9 exercise #3**

**Here we explore the maximal margin classifier on a toy data set.**

**(a) We are given $n = 7$ observations in $p = 2$ dimensions. For each observation, there is an associated class label. Sketch the observations.**

```python
import matplotlib.pyplot as plt

# Data points
red_points = [(3,4), (2,2), (4,4), (1,4)]
blue_points = [(2,1), (4,3), (4,1)]

# Plotting
plt.figure(figsize=(6, 4))
for (x, y) in red_points:
    plt.scatter(x, y, color='red', marker='o', label='Red' if (x,y)==(3,4) else "")
for (x, y) in blue_points:
    plt.scatter(x, y, color='blue', marker='s', label='Blue' if (x,y)==(2,1) else "")

plt.xlabel('X₁')
plt.ylabel('X₂')
plt.title('Observations')
plt.grid(True)
plt.legend()
plt.show()
```

**(b) Sketch the optimal separating hyperplane, and provide the equation for this hyperplane**

```python
# Plotting the hyperplane
x_vals = [1, 4]
y_vals = [x - 0.5 for x in x_vals]

plt.figure(figsize=(6, 4))
plt.plot(x_vals, y_vals, 'g--', label='Optimal Hyperplane: X₂ = X₁ - 0.5')

# Replot points
for (x, y) in red_points:
    plt.scatter(x, y, color='red', marker='o')
for (x, y) in blue_points:
    plt.scatter(x, y, color='blue', marker='s')

plt.xlabel('X₁')
plt.ylabel('X₂')
plt.title('Optimal Separating Hyperplane')
plt.grid(True)
plt.legend()
plt.show()
```

**(c) Describe the classification rule for the maximal margin classifier. It should be something along the lines of “Classify to Red if
$\beta_0 + \beta_1X_1 + \beta_2X_2 > 0$, and classify to Blue otherwise.” Provide the values for $\beta_0$, $\beta_1$, and $\beta_2$.**

The Classification rule is:
$$\text{Classify as Red if} -X_1+X_2+1 >0,\ \text{otherwise, classify as blue}$$
Coefficients:
$$\beta_0=1, \beta_1=-1, \beta_2=1$$

***

**(d) On your sketch, indicate the margin for the maximal margin hyperplane.**

```python
import numpy as np

# Margin lines (parallel to hyperplane)
margin = 1 / np.sqrt(2)  # Distance for margin

# Upper margin line: X₂ = X₁ - 1 + margin*sqrt(2)
y_upper = [x - 0.5 + margin for x in x_vals]

# Lower margin line: X₂ = X₁ - 1 - margin*sqrt(2)
y_lower = [x - 0.5 - margin for x in x_vals]

plt.figure(figsize=(6, 4))
plt.plot(x_vals, y_vals, 'g--', label='Hyperplane')
plt.plot(x_vals, y_upper, 'k:', label='Margin')
plt.plot(x_vals, y_lower, 'k:')

# Replot points
for (x, y) in red_points:
    plt.scatter(x, y, color='red', marker='o')
for (x, y) in blue_points:
    plt.scatter(x, y, color='blue', marker='s')

plt.xlabel('X₁')
plt.ylabel('X₂')
plt.title('Maximal Margin Hyperplane with Margin')
plt.grid(True)
plt.legend()
plt.show()
```

**(e) Indicate the support vectors for the maximal margin classifier.**

- Red: $(2,2), (4,4)$
- Blue: $(2,1), (4,3)$

***

**(f) Argue that a slight movement of the seventh observation would not affect the maximal margin hyperplane.**

Observation 7 $(4,1)$ is far from the hyperplane and not a support vector. Slight movement won’t affect the margin or hyperplane.

***

**(g) Sketch a hyperplane that is not the optimal separating hyperplane, and provide the equation for this hyperplane.**

```python
plt.figure(figsize=(6,4))
plt.axhline(y=2, color='purple', linestyle='-.', label='Non-Optimal Hyperplane')

# Replot points
for (x, y) in red_points:
    plt.scatter(x, y, color='red', marker='o')
for (x, y) in blue_points:
    plt.scatter(x, y, color='blue', marker='s')

plt.xlabel('X₁')
plt.ylabel('X₂')
plt.title('Non-Optimal Hyperplane')
plt.grid(True)
plt.legend()
plt.show()
```

**(h) Draw an additional observation on the plot so that the two classes are no longer separable by a hyperplane.**

```python
plt.figure(figsize=(6, 4))

x_vals = [1, 4]
y_vals = [x - 1 for x in x_vals]

plt.figure(figsize=(6, 4))
plt.plot(x_vals, y_vals, 'g--', label='Optimal Hyperplane: X₂ = X₁ - 1')
plt.scatter(3, 2.5, color='blue', marker='s', label='New Blue Point')

# Replot original points
for (x, y) in red_points:
    plt.scatter(x, y, color='red', marker='o')
for (x, y) in blue_points:
    plt.scatter(x, y, color='blue', marker='s')

plt.xlabel('X₁')
plt.ylabel('X₂')
plt.title('Non-Separable Classes')
plt.grid(True)
plt.legend()
plt.show()
```

**(b) Suppose an SVM classifier is fitted to data, resulting in the following decision rule:**
$$\hat{y}(x_1,x_2)= 
\begin{cases}
    +, & \text{if}f(x_1,x_2)>0, \\
    -, & \text{otherwise}
\end{cases}
$$
**Suppose further that this SVM is associated with a margin of m = $\sqrt{2}$. Draw the margin lines.**

```python
def hyperplane(x1):
    return x1 - np.sqrt(2)

def upper_margin(x1):
    return x1 - np.sqrt(2) - 2

def lower_margin(x1):
    return x1 - np.sqrt(2) + 2

x1 = np.linspace(-5, 5, 100)

p = [(-1, -0.5), (-2,1), (0.5,1.5), (1,-0.25), (2,0.7)] # f < 0 
n = [(-1.5, -3), (0, -2), (1,-1.5), (1/np.sqrt(2), -1), (3,1)] # f > 0

hyperplane_x2 = hyperplane(x1)
upper_x2 = upper_margin(x1)
lower_x2 = lower_margin(x1)

plt.figure(figsize=(8,4))
plt.plot(x1, hyperplane_x2, 'g--', label='Hyperplane: $X_2 = X_1 - \sqrt{2}$')
plt.plot(x1, upper_x2, 'r:', label='Upper Margin: $X_2 = X_1 - \sqrt{2} - 2$')
plt.plot(x1, lower_x2, 'b:', label='Lower Margin: $X_2 = X_1 - \sqrt{2} + 2$')

for (x,y) in n:
    plt.plot(x, y, marker="+", color="C3")
for (x,y) in p:
    plt.plot(x, y, marker="_", color="C0")

plt.fill_between(x1, upper_x2, lower_x2, color='gray', alpha=0.2, label=f'Margin ($\sqrt{{2}}$)')

plt.xlabel('$X_1$')
plt.ylabel('$X_2$')
plt.title('SVM Hyperplane with Margin')
plt.grid(True)
plt.legend()
plt.show()
```

#### Question 4

**This question uses the `crabs.csv` data, with the five body measurements as predictors and `species` as the response. Omit all other variables including `sex`. We will assess error in terms of misclassification rate.**

```python
import pandas as pd
from sklearn.metrics import accuracy_score

crabs = pd.read_csv("crabs.csv", index_col=0)
crabs.head()
y = crabs["sp"]
X = crabs.loc[:, 'FL':'BD']
```

**(a) Fit a linear SVM to the data for a range of values of the C hyperparameter, to predict `species` from the five numerical measurements. Plot the cross-validated estimate of the error as a function of C.**

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# Encode the target labels (species) if necessary
le = LabelEncoder()
y = le.fit_transform(y)

C_values = np.logspace(-3, 3, 7)  # C values from 0.001 to 1000
linear_errors = []

for C in C_values:
    svm = SVC(kernel='linear', C=C)
    error = 1 - cross_val_score(svm, X, y, cv=5, scoring='accuracy').mean()  # Misclassification error
    linear_errors.append(error)

# Plot the error for linear SVM
plt.figure(figsize=(8, 5))
plt.plot(C_values, linear_errors, marker='o', label='Linear SVM')
plt.xscale('log')
plt.xlabel('C')
plt.ylabel('Cross-validated Misclassification Error')
plt.title('Linear SVM: Misclassification Error vs C')
plt.grid(True)
plt.legend()
plt.show()
```

**(b) Now, fit nonlinear SVMs with polynomial kernels. Consider at least three values of degree. For each, plot the cross-validated estimate of the error as a function of C.**

```python
degrees = [2, 3, 4]
C_values = np.logspace(-3, 3, 7)  # C values from 0.001 to 1000
polynomial_errors = {degree: [] for degree in degrees}

for degree in degrees:
    for C in C_values:
        svm = SVC(kernel='poly', C=C, degree=degree)
        error = 1 - cross_val_score(svm, X, y, cv=5, scoring='accuracy').mean()  # Misclassification error
        polynomial_errors[degree].append(error)

plt.figure(figsize=(10, 6))
for degree, errors in polynomial_errors.items():
    plt.plot(C_values, errors, marker='o', label=f'Polynomial Degree {degree}')
plt.xscale('log')
plt.xlabel('C')
plt.ylabel('Cross-validated Misclassification Error')
plt.title('Polynomial Kernel SVM: Misclassification Error vs C (Degree)')
plt.grid(True)
plt.legend()
plt.show()
```

**(c) Finally, fit nonlinear SVMs with radial kernels. Consider at least three values of gamma. For each, plot the cross-validated estimate of the error as a function of C.**

```python
gamma_values = [0.1, 1, 10]  # Gamma values to test
rbf_errors = {gamma: [] for gamma in gamma_values}
C_values = np.logspace(-3, 3, 7)

for gamma in gamma_values:
    for C in C_values:
        svm = SVC(kernel='rbf', gamma=gamma, C=C)
        error = 1 - cross_val_score(svm, X, y, cv=5, scoring='accuracy').mean()
        rbf_errors[gamma].append(error)

plt.figure(figsize=(10, 6))
for gamma, errors in rbf_errors.items():
    plt.plot(C_values, errors, marker='o', label=f'Gamma Value {gamma}')
plt.xscale('log')
plt.xlabel('C')
plt.ylabel('Cross-validated Misclassification Error')
plt.title('Radial Kernel SVM: Misclassification Error vs C (gamma)')
plt.grid(True)
plt.legend()
plt.show()
```
